```python
def missing_report(df):
    report = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": df.isnull().mean() * 100
    })

    return report.sort_values("Missing Percentage", ascending=False)
```

```python
def datatype_report(df):
    return pd.DataFrame({
        "Data Type": df.dtypes,
        "Non-Null Count": df.notnull().sum(),
        "Unique Values": df.nunique()
    })

display(datatype_report(crop_clean))
display(datatype_report(train_clean))
display(datatype_report(test_clean))
```

```python
def infinite_report(df):
    numeric_df = df.select_dtypes(include=np.number)

    return pd.DataFrame({
        "Positive Infinity": np.isposinf(numeric_df).sum(),
        "Negative Infinity": np.isneginf(numeric_df).sum()
    })
```

```python
print("CROP RECOMMENDATION")
display(infinite_report(crop_clean))

print("AGRIYIELD TRAIN")
display(infinite_report(train_clean))

print("AGRIYIELD TEST")
display(infinite_report(test_clean))
```

```python
def numerical_summary(df):
    numeric_df = df.select_dtypes(include=np.number)

    return numeric_df.describe().T[
        ["min", "max", "mean", "std"]
    ]
```

```python
print("CROP RECOMMENDATION")
display(numerical_summary(crop_clean))

print("AGRIYIELD TRAIN")
display(numerical_summary(train_clean))

print("AGRIYIELD TEST")
display(numerical_summary(test_clean))
```

```python
def categorical_report(df):
    categorical_cols = df.select_dtypes(
        include=["object", "category"]
    ).columns

    for col in categorical_cols:
        print("\n" + "=" * 60)
        print("COLUMN:", col)
        print("Unique values:", df[col].nunique(dropna=False))
        print(df[col].value_counts(dropna=False))
```

```python
def outlier_report(df):
    numeric_df = df.select_dtypes(include=np.number)

    results = []

    for col in numeric_df.columns:

        Q1 = numeric_df[col].quantile(0.25)
        Q3 = numeric_df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        count = (
            (numeric_df[col] < lower) |
            (numeric_df[col] > upper)
        ).sum()

        percentage = count / len(df) * 100

        results.append([
            col,
            Q1,
            Q3,
            lower,
            upper,
            count,
            percentage
        ])

    return pd.DataFrame(
        results,
        columns=[
            "Column",
            "Q1",
            "Q3",
            "Lower Bound",
            "Upper Bound",
            "Outlier Count",
            "Outlier Percentage"
        ]
    ).sort_values(
        "Outlier Percentage",
        ascending=False
    )
```

```python
print("CROP RECOMMENDATION OUTLIERS")
display(outlier_report(crop_clean))

print("AGRIYIELD TRAIN OUTLIERS")
display(outlier_report(train_clean))

print("AGRIYIELD TEST OUTLIERS")
display(outlier_report(test_clean))
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("agriculture.db")

crop_clean = pd.read_csv("Crop_recommendation_clean.csv")
train_clean = pd.read_csv("train_clean.csv")
test_clean = pd.read_csv("test_clean.csv")

crop_clean.to_sql(
    "crop_recommendation_clean",
    conn,
    if_exists="replace",
    index=False
)

train_clean.to_sql(
    "agri_yield_train_clean",
    conn,
    if_exists="replace",
    index=False
)

test_clean.to_sql(
    "agri_yield_test_clean",
    conn,
    if_exists="replace",
    index=False
)

conn.close()

print("Clean datasets successfully loaded into SQLite.")
```
