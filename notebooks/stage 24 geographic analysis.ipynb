```python
import sqlite3
import pandas as pd
import os

os.makedirs("stage24_outputs", exist_ok=True)

db_path = r"C:\Users\ADMIN\smart agricultural analytics\agriculture.db"
conn = sqlite3.connect(db_path)

tables = [
    "crop_recommendation_selected",
    "agri_yield_train_selected",
    "agri_yield_test_selected"
]

for table in tables:
    columns = pd.read_sql(
        f"PRAGMA table_info({table})", conn
    )["name"].tolist()

    print(f"\n{table}:")
    print(columns)

conn.close()
```
