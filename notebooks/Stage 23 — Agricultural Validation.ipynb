```python
import pandas as pd
import sqlite3
import os

os.makedirs("stage23_outputs", exist_ok=True)

conn = sqlite3.connect(
    r"C:\Users\ADMIN\smart agricultural analytics\agriculture.db"
)

crop = pd.read_sql("SELECT * FROM crop_recommendation_selected", conn)
yield_df = pd.read_sql("SELECT * FROM agri_yield_train_selected", conn)

conn.close()

print("Crop data:", crop.shape)
print("Yield data:", yield_df.shape)
```

```python
print(crop[["temperature", "humidity", "rainfall"]].describe())

print("\nCrop class counts:")
print(crop["label"].value_counts())

crop["label"].value_counts().to_csv(
    "stage23_outputs/crop_class_counts.csv"
)
```

```python
print(yield_df["yield"].describe())

features = [
    "NDVI_Temperature", "OrganicMatter_Sand",
    "Temperature_Rainfall", "sand_pct", "NDVI_Rainfall",
    "soil_ph", "organic_matter", "Humidity_Rainfall",
    "Temperature_Humidity", "rainfall"
]

corr = yield_df[features + ["yield"]].corr()["yield"].drop("yield")

print("\nFeature correlations with yield:")
print(corr.sort_values(ascending=False))

corr.to_csv("stage23_outputs/yield_correlations.csv")
```
