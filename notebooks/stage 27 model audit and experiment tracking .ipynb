```python
import sqlite3
import pandas as pd
import os

db = r"C:\Users\ADMIN\smart agricultural analytics\agriculture.db"
os.makedirs("stage27_outputs", exist_ok=True)

conn = sqlite3.connect(db)

tables = [
    ("crop_recommendation_fe", "crop_recommendation_selected"),
    ("agri_yield_train_fe", "agri_yield_train_selected")
]

results = []

for original, selected in tables:
    a = pd.read_sql(f"SELECT * FROM {original} LIMIT 1", conn)
    b = pd.read_sql(f"SELECT * FROM {selected} LIMIT 1", conn)

    results.append({
        "Selected Table": selected,
        "Missing Columns": str(
            list(set(b.columns) - set(a.columns))
        )
    })

conn.close()

pd.DataFrame(results).to_csv(
    "stage27_outputs/feature_audit.csv", index=False
)

print(pd.DataFrame(results))
print("Audit saved!")
```

```python
import sqlite3
import pandas as pd
import joblib
from sklearn.metrics import accuracy_score, classification_report

db_path = r"C:\Users\ADMIN\smart agricultural analytics\agriculture.db"

conn = sqlite3.connect(db_path)
df = pd.read_sql("SELECT * FROM crop_recommendation_selected", conn)
conn.close()

features = [
    "humidity", "K", "temperature", "N", "P_K_Ratio",
    "NPK_Total", "Temperature_Rainfall", "rainfall",
    "N_K_Ratio", "Humidity_Rainfall",
    "Temperature_Humidity", "P"
]

X = df[features]
y = df["label"]

model = joblib.load("stage26_models/crop_classifier.joblib")
predictions = model.predict(X)

print("Accuracy:", accuracy_score(y, predictions))
print("\nClassification Report:\n")
print(classification_report(y, predictions))
```
