```python
import sqlite3
import pandas as pd
import os

os.makedirs("stage25_outputs", exist_ok=True)

conn = sqlite3.connect(
    r"C:\Users\ADMIN\smart agricultural analytics\agriculture.db"
)

df = pd.read_sql("SELECT * FROM crop_recommendation", conn)
conn.close()

print(df.shape)
```

```python
features = [
    "pest_pressure", "frost_risk", "soil_moisture",
    "water_usage_efficiency", "irrigation_frequency"
]

for col in features:
    print("\n", col)
    print(df[col].value_counts(dropna=False).head(10))
```

```python
results = []

for col in features:
    results.append({
        "Feature": col,
        "Missing": df[col].isna().sum(),
        "Unique_Values": df[col].nunique()
    })

pd.DataFrame(results).to_csv(
    "stage25_outputs/risk_summary.csv", index=False
)

print("Risk analysis completed.")
```
